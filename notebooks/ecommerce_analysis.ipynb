```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

print(" Starting E-Commerce Analysis with Correct Datasets...")

# Load the CORRECT datasets
print("📊 Loading correct datasets...")
olist_orders = pd.read_csv('olist_orders_dataset.csv')  # This has delivery dates
olist_order_items = pd.read_csv('olist_order_items_dataset.csv')

print(f" Orders dataset shape: {olist_orders.shape}")
print(f"Order items dataset shape: {olist_order_items.shape}")

print("\n Orders dataset columns:")
for col in olist_orders.columns:
    print(f"    {col}")

print("\n Order items dataset columns:")
for col in olist_order_items.columns:
    print(f"    {col}")

# Merge the correct datasets
print("\n Merging datasets...")
combined_orders_dataset = pd.merge(
    olist_orders,
    olist_order_items,
    on='order_id',
    how='inner'
)

print(f" Merged dataset shape: {combined_orders_dataset.shape}")
print("\n Combined dataset columns:")
for col in combined_orders_dataset.columns:
    print(f"    {col}")

# Now we should have the correct date columns
print("\n Converting date columns...")
combined_orders_dataset['order_purchase_timestamp'] = pd.to_datetime(combined_orders_dataset['order_purchase_timestamp'])
combined_orders_dataset['order_delivered_customer_date'] = pd.to_datetime(combined_orders_dataset['order_delivered_customer_date'])

print("Date conversion successful!")

# Calculate delivery time in days
print("\n📅 Calculating delivery times...")
combined_orders_dataset['delivery_time_days'] = (
    combined_orders_dataset['order_delivered_customer_date'] - 
    combined_orders_dataset['order_purchase_timestamp']
).dt.days

# Remove invalid delivery times
valid_deliveries = combined_orders_dataset[
    (combined_orders_dataset['delivery_time_days'] >= 0) & 
    (combined_orders_dataset['delivery_time_days'] <= 365)
]

print(f"Valid deliveries for analysis: {len(valid_deliveries):,}")

# Create delivery time histogram
print("\n📊 Creating visualization...")
plt.figure(figsize=(12, 6))
n, bins, patches = plt.hist(valid_deliveries['delivery_time_days'], bins=50, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Distribution of Delivery Times - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Delivery Time (Days)', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)
plt.grid(axis='y', alpha=0.3)

# Add mean and median lines
avg_delivery = valid_deliveries['delivery_time_days'].mean()
median_delivery = valid_deliveries['delivery_time_days'].median()

plt.axvline(avg_delivery, color='red', linestyle='--', linewidth=2, label=f'Mean: {avg_delivery:.1f} days')
plt.axvline(median_delivery, color='green', linestyle='--', linewidth=2, label=f'Median: {median_delivery:.1f} days')
plt.legend()

# Save the plot for your portfolio
plt.savefig('delivery_time_distribution.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# Delivery time metrics
print("\n📈 DELIVERY PERFORMANCE METRICS:")
print("=" * 45)
print(f"📦 Average delivery time: {avg_delivery:.2f} days")
print(f"📦 Median delivery time: {median_delivery:.2f} days")
print(f"📦 Fastest delivery: {valid_deliveries['delivery_time_days'].min()} days")
print(f"📦 Slowest delivery: {valid_deliveries['delivery_time_days'].max()} days")
print(f"📦 Standard deviation: {valid_deliveries['delivery_time_days'].std():.2f} days")

# Delivery performance categories
fast_delivery = valid_deliveries[valid_deliveries['delivery_time_days'] <= 7]
medium_delivery = valid_deliveries[(valid_deliveries['delivery_time_days'] > 7) & (valid_deliveries['delivery_time_days'] <= 30)]
slow_delivery = valid_deliveries[valid_deliveries['delivery_time_days'] > 30]

print(f"\n Fast delivery (≤7 days): {len(fast_delivery):,} orders ({len(fast_delivery)/len(valid_deliveries)*100:.1f}%)")
print(f" Medium delivery (8-30 days): {len(medium_delivery):,} orders ({len(medium_delivery)/len(valid_deliveries)*100:.1f}%)")
print(f" Slow delivery (>30 days): {len(slow_delivery):,} orders ({len(slow_delivery)/len(valid_deliveries)*100:.1f}%)")

print("\n Delivery analysis completed successfully!")

#Top Selling Products Analysis

print("\n" + "="*50)
print("Top Selling Products Analysis")
print("="*50)

# 1. Identify the top 10 most sold products
top_products = (combined_orders_dataset
    .groupby('product_id')
    .agg({
        'order_item_id': 'count',
        'price': 'mean'
    })
    .rename(columns={'order_item_id': 'quantity_sold', 'price': 'avg_price'})
    .sort_values('quantity_sold', ascending=False)
    .head(10)
    .reset_index()
)

print("\n📊 Top 10 Most Sold Products:")
print(top_products[['product_id', 'quantity_sold', 'avg_price']].round(2))

# 2. Create bar chart
plt.figure(figsize=(14, 8))
bars = plt.barh(
    range(len(top_products)),
    top_products['quantity_sold'],
    color='lightcoral',
    alpha=0.7,
    edgecolor='darkred'
)

plt.title('Top 10 Most Sold Products - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Quantity Sold', fontsize=12)
plt.ylabel('Product ID', fontsize=12)
plt.yticks(range(len(top_products)), top_products['product_id'])

# Add value labels on bars
for bar, value in zip(bars, top_products['quantity_sold']):
    plt.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{int(value)}', ha='left', va='center', fontweight='bold')

plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('top_products_chart.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# Seller Revenue Analysis

print("\n" + "="*50)
print("Seller Revenue Analysis")
print("="*50)

# 1. Calculate total revenue (product price + freight value) for each seller
seller_revenue = (combined_orders_dataset
    .groupby('seller_id')
    .agg({
        'price': 'sum',
        'freight_value': 'sum',
        'order_id': 'count'  # Count number of orders
    })
    .reset_index()
)

seller_revenue['total_revenue'] = seller_revenue['price'] + seller_revenue['freight_value']
seller_revenue = seller_revenue.rename(columns={'order_id': 'order_count'})

# Get top 10 sellers by revenue
top_sellers = seller_revenue.nlargest(10, 'total_revenue')

print("\n📊 Top 10 Sellers by Revenue:")
print(top_sellers[['seller_id', 'total_revenue', 'order_count', 'price', 'freight_value']].round(2))

# 2. Create bar chart for top sellers
plt.figure(figsize=(14, 8))
bars = plt.barh(
    range(len(top_sellers)),
    top_sellers['total_revenue'],
    color='lightgreen',
    alpha=0.7,
    edgecolor='darkgreen'
)

plt.title('Top 10 Sellers by Total Revenue - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Total Revenue (R$)', fontsize=12)
plt.ylabel('Seller ID', fontsize=12)
plt.yticks(range(len(top_sellers)), top_sellers['seller_id'])

# Add value labels on bars
for bar, value in zip(bars, top_sellers['total_revenue']):
    plt.text(bar.get_width() + 100, bar.get_y() + bar.get_height()/2, 
             f'R$ {value:,.0f}', ha='left', va='center', fontweight='bold')

plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('seller_revenue_chart.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 3. Revenue breakdown: Price vs Freight
plt.figure(figsize=(14, 8))

# Create stacked bar chart
bar_width = 0.6
bars1 = plt.barh(
    range(len(top_sellers)),
    top_sellers['price'],
    bar_width,
    color='lightblue',
    label='Product Price',
    alpha=0.7
)

bars2 = plt.barh(
    range(len(top_sellers)),
    top_sellers['freight_value'],
    bar_width,
    left=top_sellers['price'],
    color='orange',
    label='Freight Value',
    alpha=0.7
)

plt.title('Top 10 Sellers Revenue Breakdown (Price vs Freight) - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Revenue (R$)', fontsize=12)
plt.ylabel('Seller ID', fontsize=12)
plt.yticks(range(len(top_sellers)), top_sellers['seller_id'])
plt.legend()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('seller_revenue_breakdown.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 4. Advanced Seller Analysis
print("\n" + "="*40)
print("SELLER PERFORMANCE ANALYSIS")
print("="*40)

# Calculate percentage of total revenue from top sellers
total_revenue_all_sellers = seller_revenue['total_revenue'].sum()
top_10_revenue = top_sellers['total_revenue'].sum()
percentage_revenue_top_10 = (top_10_revenue / total_revenue_all_sellers) * 100

# Calculate average order value for top sellers
top_sellers['avg_order_value'] = top_sellers['total_revenue'] / top_sellers['order_count']
top_sellers['freight_percentage'] = (top_sellers['freight_value'] / top_sellers['total_revenue']) * 100

print(f"💰 Total revenue from all sellers: R$ {total_revenue_all_sellers:,.2f}")
print(f"💰 Revenue from top 10 sellers: R$ {top_10_revenue:,.2f}")
print(f"💰 Top 10 sellers account for: {percentage_revenue_top_10:.2f}% of total revenue")

print(f"\n📦 Average Order Value for Top Sellers: R$ {top_sellers['avg_order_value'].mean():.2f}")
print(f"🚚 Average freight as % of revenue: {top_sellers['freight_percentage'].mean():.2f}%")

# Seller concentration analysis
print(f"\n📊 Seller Distribution Analysis:")
print(f"   Total number of sellers: {len(seller_revenue)}")
print(f"   Average revenue per seller: R$ {seller_revenue['total_revenue'].mean():.2f}")
print(f"   Median revenue per seller: R$ {seller_revenue['total_revenue'].median():.2f}")

# Identify revenue concentration
top_1_percent = max(1, int(len(seller_revenue) * 0.01))
top_1_percent_revenue = seller_revenue.nlargest(top_1_percent, 'total_revenue')['total_revenue'].sum()
percentage_top_1 = (top_1_percent_revenue / total_revenue_all_sellers) * 100

print(f"   Top 1% of sellers account for: {percentage_top_1:.2f}% of total revenue")

print("\n Seller revenue analysis completed successfully!")

# Monthly Revenue Trends Analysis

print("\n" + "="*50)
print("Monthly Revenue Trends Analysis")
print("="*50)

# 1. Group data by month and calculate total revenue
monthly_revenue = combined_orders_dataset.copy()

# Calculate total revenue per order item
monthly_revenue['total_revenue'] = monthly_revenue['price'] + monthly_revenue['freight_value']

# Extract year-month from purchase timestamp
monthly_revenue['order_month'] = monthly_revenue['order_purchase_timestamp'].dt.to_period('M')

# Group by month and calculate metrics
monthly_totals = (monthly_revenue
    .groupby('order_month')
    .agg({
        'total_revenue': 'sum',
        'order_id': 'nunique',  # Count unique orders
        'product_id': 'count'   # Count total items sold
    })
    .reset_index()
    .rename(columns={
        'order_id': 'order_count',
        'product_id': 'item_count'
    })
)

# Convert period to timestamp for plotting
monthly_totals['order_month'] = monthly_totals['order_month'].dt.to_timestamp()

# Sort by month to ensure proper line chart
monthly_totals = monthly_totals.sort_values('order_month')

print("\n📅 Monthly Revenue Summary:")
print(monthly_totals[['order_month', 'total_revenue', 'order_count', 'item_count']].round(2))

# 2. Create line chart for monthly revenue trends
plt.figure(figsize=(15, 8))

# Main revenue trend line
plt.plot(monthly_totals['order_month'], 
         monthly_totals['total_revenue'], 
         marker='o', 
         linewidth=3, 
         markersize=8,
         color='royalblue',
         label='Total Revenue')

# Customize the chart
plt.title('Monthly Revenue Trends - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Revenue (R$)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)

# Add value annotations for peaks and valleys
max_revenue = monthly_totals['total_revenue'].max()
min_revenue = monthly_totals['total_revenue'].min()
max_month = monthly_totals.loc[monthly_totals['total_revenue'].idxmax(), 'order_month']
min_month = monthly_totals.loc[monthly_totals['total_revenue'].idxmin(), 'order_month']

plt.annotate(f'Peak: R$ {max_revenue:,.0f}', 
             xy=(max_month, max_revenue), 
             xytext=(10, 20), 
             textcoords='offset points',
             arrowprops=dict(arrowstyle='->', color='red'),
             fontweight='bold',
             color='red')

plt.annotate(f'Low: R$ {min_revenue:,.0f}', 
             xy=(min_month, min_revenue), 
             xytext=(10, -30), 
             textcoords='offset points',
             arrowprops=dict(arrowstyle='->', color='green'),
             fontweight='bold',
             color='green')

plt.legend()
plt.tight_layout()
plt.savefig('monthly_revenue_trends.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 3. Multiple metrics view
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(15, 15))

# Subplot 1: Revenue trend
ax1.plot(monthly_totals['order_month'], monthly_totals['total_revenue'], 
         marker='o', linewidth=2, color='royalblue')
ax1.set_title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
ax1.set_ylabel('Revenue (R$)', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# Subplot 2: Order count trend
ax2.plot(monthly_totals['order_month'], monthly_totals['order_count'], 
         marker='s', linewidth=2, color='coral')
ax2.set_title('Monthly Order Count Trend', fontsize=14, fontweight='bold')
ax2.set_ylabel('Number of Orders', fontsize=12)
ax2.grid(True, alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# Subplot 3: Average order value trend
monthly_totals['avg_order_value'] = monthly_totals['total_revenue'] / monthly_totals['order_count']
ax3.plot(monthly_totals['order_month'], monthly_totals['avg_order_value'], 
         marker='^', linewidth=2, color='green')
ax3.set_title('Monthly Average Order Value Trend', fontsize=14, fontweight='bold')
ax3.set_xlabel('Month', fontsize=12)
ax3.set_ylabel('Average Order Value (R$)', fontsize=12)
ax3.grid(True, alpha=0.3)
ax3.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('monthly_trends_comprehensive.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 4. Advanced Trend Analysis
print("\n" + "="*40)
print("TREND ANALYSIS & INSIGHTS")
print("="*40)

# Calculate growth metrics
monthly_totals['revenue_growth'] = monthly_totals['total_revenue'].pct_change() * 100

# Identify top and bottom performing months
top_3_months = monthly_totals.nlargest(3, 'total_revenue')
bottom_3_months = monthly_totals.nsmallest(3, 'total_revenue')

print("\n Top 3 Highest Revenue Months:")
for _, row in top_3_months.iterrows():
    month_str = row['order_month'].strftime('%B %Y')
    print(f"   {month_str}: R$ {row['total_revenue']:,.2f} ({row['order_count']} orders)")

print("\n📉 Bottom 3 Lowest Revenue Months:")
for _, row in bottom_3_months.iterrows():
    month_str = row['order_month'].strftime('%B %Y')
    print(f"   {month_str}: R$ {row['total_revenue']:,.2f} ({row['order_count']} orders)")

# Seasonal analysis
print("\n" + "="*40)
print("SEASONAL PATTERNS ANALYSIS")
print("="*40)

print("\n📈 Key Business Insights:")
print("• GROWTH TREND: Revenue shows " + ("📈 upward" if monthly_totals['revenue_growth'].mean() > 0 else "📉 downward") + " trend")
print("• SEASONALITY: Look for repeating patterns around holidays and special events")
print("• ORDER VALUE: Average order value is " + ("increasing" if monthly_totals['avg_order_value'].iloc[-1] > monthly_totals['avg_order_value'].iloc[0] else "decreasing"))

# Statistical summary
print("\n" + "="*40)
print("STATISTICAL SUMMARY")
print("="*40)

print(f"📊 Average Monthly Revenue: R$ {monthly_totals['total_revenue'].mean():,.2f}")
print(f"📊 Highest Monthly Revenue: R$ {monthly_totals['total_revenue'].max():,.2f}")
print(f"📊 Lowest Monthly Revenue: R$ {monthly_totals['total_revenue'].min():,.2f}")
print(f"📊 Revenue Volatility (Std Dev): R$ {monthly_totals['total_revenue'].std():,.2f}")

print(f"\n📦 Average Monthly Orders: {monthly_totals['order_count'].mean():.0f}")
print(f" Average Order Value: R$ {monthly_totals['avg_order_value'].mean():.2f}")

# Growth analysis
positive_growth_months = monthly_totals[monthly_totals['revenue_growth'] > 0]
if len(positive_growth_months) > 0:
    avg_positive_growth = positive_growth_months['revenue_growth'].mean()
    print(f"📈 Average positive monthly growth: {avg_positive_growth:.2f}%")

# 5. Business Recommendations
print("\n" + "="*40)
print("DATA-DRIVEN RECOMMENDATIONS")
print("="*40)

print("\n Marketing & Business Strategies:")

# Identify low-performing months for targeted strategies
low_months = bottom_3_months['order_month'].dt.strftime('%B').tolist()
if low_months:
    unique_low_months = list(set(low_months))
    print(f"1. FOCUS ON SLOW PERIODS: Boost marketing during {', '.join(unique_low_months)}")

print("2. CAPITALIZE ON PEAKS: Increase inventory and staffing during high-revenue months")
print("3. CUSTOMER RETENTION: Implement loyalty programs to maintain consistent revenue")
print("4. CROSS-SELLING: Use top products to drive sales of complementary items")
print("5. SELLER PARTNERSHIPS: Strengthen relationships with top-performing sellers")
print("6. SEASONAL PLANNING: Align marketing campaigns with identified seasonal patterns")
print("7. OPERATIONAL EFFICIENCY: Optimize delivery times to improve customer satisfaction")
print("8. PRICING STRATEGY: Monitor average order value trends for pricing optimization")

print("\n Monthly trends analysis completed successfully!")
print("\n E-commerce analysis finished.")

# Save the final combined dataset for future use
combined_orders_dataset.to_csv('final_analyzed_dataset.csv', index=False)
print("\n Final analyzed dataset saved as 'final_analyzed_dataset.csv'")

# Create Dashboard Preview Image
print("\n" + "="*50)
print("Creating Dashboard Preview for Portfolio")
print("="*50)

# Create a comprehensive dashboard preview
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(20, 12))
fig.suptitle('E-Commerce Analytics Dashboard - Olist Dataset Analysis', fontsize=20, fontweight='bold', y=0.98)

# 1. Delivery Time Distribution
ax1.hist(valid_deliveries['delivery_time_days'], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
ax1.axvline(avg_delivery, color='red', linestyle='--', linewidth=2, label=f'Mean: {avg_delivery:.1f} days')
ax1.axvline(median_delivery, color='green', linestyle='--', linewidth=2, label=f'Median: {median_delivery:.1f} days')
ax1.set_title('Delivery Time Distribution', fontsize=14, fontweight='bold')
ax1.set_xlabel('Delivery Time (Days)')
ax1.set_ylabel('Number of Orders')
ax1.legend()
ax1.grid(alpha=0.3)

# 2. Top Products (simplified for dashboard)
top_5_products = top_products.head(5)
y_pos = range(len(top_5_products))
ax2.barh(y_pos, top_5_products['quantity_sold'], color='lightcoral', alpha=0.7, edgecolor='darkred')
ax2.set_yticks(y_pos)
ax2.set_yticklabels([f'Product {i+1}' for i in range(len(top_5_products))])
ax2.set_title('Top 5 Selling Products', fontsize=14, fontweight='bold')
ax2.set_xlabel('Quantity Sold')
ax2.grid(axis='x', alpha=0.3)

# Add value labels
for i, v in enumerate(top_5_products['quantity_sold']):
    ax2.text(v + 1, i, str(int(v)), va='center', fontweight='bold')

# 3. Top Sellers Revenue
top_5_sellers = top_sellers.head(5)
y_pos_sellers = range(len(top_5_sellers))
bar_width = 0.6
ax3.barh(y_pos_sellers, top_5_sellers['price'], bar_width, color='lightblue', label='Product Price', alpha=0.7)
ax3.barh(y_pos_sellers, top_5_sellers['freight_value'], bar_width, left=top_5_sellers['price'], 
         color='orange', label='Freight Value', alpha=0.7)
ax3.set_yticks(y_pos_sellers)
ax3.set_yticklabels([f'Seller {i+1}' for i in range(len(top_5_sellers))])
ax3.set_title('Top 5 Sellers Revenue Breakdown', fontsize=14, fontweight='bold')
ax3.set_xlabel('Revenue (R$)')
ax3.legend()
ax3.grid(axis='x', alpha=0.3)

# 4. Monthly Revenue Trend
ax4.plot(monthly_totals['order_month'], monthly_totals['total_revenue'], 
         marker='o', linewidth=2, color='royalblue')
ax4.set_title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
ax4.set_xlabel('Month')
ax4.set_ylabel('Revenue (R$)')
ax4.tick_params(axis='x', rotation=45)
ax4.grid(alpha=0.3)

# Add key metrics as text
metrics_text = f"""
Key Metrics:
• Avg Delivery: {avg_delivery:.1f} days
• Fast Delivery (≤7 days): {len(fast_delivery)/len(valid_deliveries)*100:.1f}%
• Top 10 Sellers: {percentage_revenue_top_10:.1f}% of revenue
• Peak Revenue: R$ {max_revenue:,.0f}
• Avg Order Value: R$ {monthly_totals['avg_order_value'].mean():.2f}
"""

fig.text(0.02, 0.02, metrics_text, fontsize=12, verticalalignment='bottom',
         bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgray", alpha=0.7))

plt.tight_layout()
plt.savefig('dashboard_preview.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("Dashboard preview created: 'dashboard_preview.png'")

# Enhanced individual charts with consistent styling
def apply_dashboard_style():
    """Apply consistent styling to all charts"""
    plt.rcParams['figure.figsize'] = (12, 6)
    plt.rcParams['font.size'] = 12
    plt.rcParams['axes.grid'] = True
    plt.rcParams['grid.alpha'] = 0.3

# Recreate main charts with better styling
apply_dashboard_style()

# 1. Enhanced Delivery Time Chart
plt.figure(figsize=(14, 8))
n, bins, patches = plt.hist(valid_deliveries['delivery_time_days'], bins=50, alpha=0.7, 
                           color='#1f77b4', edgecolor='black')
plt.title('Distribution of Delivery Times - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Delivery Time (Days)', fontsize=14)
plt.ylabel('Number of Orders', fontsize=14)
plt.axvline(avg_delivery, color='red', linestyle='--', linewidth=2, label=f'Mean: {avg_delivery:.1f} days')
plt.axvline(median_delivery, color='green', linestyle='--', linewidth=2, label=f'Median: {median_delivery:.1f} days')
plt.legend(fontsize=12)
plt.grid(alpha=0.3)
plt.savefig('delivery_time_distribution_enhanced.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 2. Enhanced Top Products Chart
plt.figure(figsize=(14, 10))
top_10_display = top_products.head(10)
colors = plt.cm.Set3(np.linspace(0, 1, len(top_10_display)))
bars = plt.barh(range(len(top_10_display)), top_10_display['quantity_sold'], color=colors, alpha=0.8)

plt.title('Top 10 Most Sold Products - Olist E-Commerce', fontsize=16, fontweight='bold')
plt.xlabel('Quantity Sold', fontsize=14)
plt.ylabel('Product ID', fontsize=14)
plt.yticks(range(len(top_10_display)), top_10_display['product_id'])

# Add value labels and prices
for i, (bar, qty, price) in enumerate(zip(bars, top_10_display['quantity_sold'], top_10_display['avg_price'])):
    plt.text(bar.get_width() + 2, bar.get_y() + bar.get_height()/2, 
             f'{int(qty)} units\nR$ {price:.2f}', 
             ha='left', va='center', fontweight='bold', fontsize=10)

plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('top_products_chart_enhanced.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("Enhanced individual charts created!")
```
